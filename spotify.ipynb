```python
# 4. Standardize dimensional ranges to identical scales (0 to 1)
scaler = MinMaxScaler()
normalized_features = scaler.fit_transform(df[feature_cols])
```

```python
#matrix
# 5. Compute the global linear cosine similarity matrix
similarity_matrix = cosine_similarity(normalized_features)
similarity_matrix
```

```python
def get_recommendations(track_name, df, similarity_matrix, top_n=5):
    # 6. Search for the track index inside the dataframe
    track_name_lower = track_name.lower()
    matches = df[df['track_name'].str.lower() == track_name_lower]
    
    if matches.empty:
        return f"Track '{track_name}' not found in the dataset database."
    
    idx = matches.index[0]
    
    # 7. Extract item likeness scores and sort descending
    sim_scores = list(enumerate(similarity_matrix[idx]))
    sim_scores = sorted(sim_scores, key=lambda x: x[1], reverse=True)
    
    # Omit the first item since it maps directly back to itself
    top_indices = [i[0] for i in sim_scores[1:top_n+1]]
    
    # 8. Filter and isolate specific visualization output fields
    recommendations = df.iloc[top_indices][['track_name', 'artists', 'album_name', 'popularity']]
    return recommendations
```
